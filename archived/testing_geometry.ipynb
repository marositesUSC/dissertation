```python
def create_data(max_range=750, bin_width=2.5, origin=[50,25,2], orgin_Azimuth=0, azimuth=0, zenith=0):
    # Create initial pandas df
    data_range = np.arange(bin_width, max_range+bin_width, bin_width)    
    data = pd.DataFrame(data_range, columns=['distance'])

    origin_X = origin[0]
    origin_Y = origin[1]
    origin_Z = origin[2]

    data['x'] = origin_X + data['distance'] * np.cos(np.radians(zenith)) * np.sin(np.radians(azimuth+orgin_Azimuth))
    data['y'] = origin_Y + data['distance'] * np.cos(np.radians(zenith)) * np.cos(np.radians(azimuth+orgin_Azimuth))
    data['z'] = origin_Z + data['distance'] * np.sin(np.radians(zenith))

    return data

```

```python
usc_0 = create_data(origin=[50,100, 0], orgin_Azimuth=110, azimuth=0, zenith=0)
usc_1 = create_data(origin=[50,100, 0], orgin_Azimuth=110, azimuth=0, zenith=1)
usc_2 = create_data(origin=[50,100, 0], orgin_Azimuth=110, azimuth=0, zenith=2)
usc_3 = create_data(origin=[50,100, 0], orgin_Azimuth=110, azimuth=0, zenith=3)
usc_4 = create_data(origin=[50,100, 0], orgin_Azimuth=110, azimuth=0, zenith=4)
usc_5 = create_data(origin=[50,100, 0], orgin_Azimuth=110, azimuth=0, zenith=5)
```

```python
uiuc_0 = create_data(origin=[300,-100, 0], orgin_Azimuth=294, azimuth=110, zenith=0)
uiuc_1 = create_data(origin=[300,-100, 0], orgin_Azimuth=294, azimuth=25, zenith=1)
uiuc_2 = create_data(origin=[300,-100, 0], orgin_Azimuth=294, azimuth=25, zenith=2)
uiuc_3 = create_data(origin=[300,-100, 0], orgin_Azimuth=294, azimuth=25, zenith=3)
uiuc_4 = create_data(origin=[300,-100, 0], orgin_Azimuth=294, azimuth=25, zenith=4)
uiuc_5 = create_data(origin=[300,-100, 0], orgin_Azimuth=294, azimuth=25, zenith=5)
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

for d in [usc_0, usc_1, usc_2, usc_3, usc_4, usc_5]:
    ax.scatter(d['x'], d['y'], d['z'])

for d in [uiuc_0, uiuc_1, uiuc_2, uiuc_3, uiuc_4, uiuc_5]:
    ax.scatter(d['x'], d['y'], d['z'])
    
ax.set_ylabel('Northing')
ax.set_xlabel('Easting')
ax.set_zlabel('Elevation')

# ax.set_xlim(350,400)
# ax.set_ylim(-50,50)
# ax.set_zlim(-200,200)


plt.show()
```
