## Decided just to move the data. 

I will use `shutil`.

```python
import shutil

for f in files:
    r = f[0]
    yyyy = '20'+f[1:3]
    month = f[3]
    if month == 'A':
        mm = '10'
    elif month == 'B':
        mm = '11'
    else:
        mm = '09'
    dd = f[4:6]
    hh = f[6:8]
    minutes = f[9:11]
    seconds = f[12:]
    iop = f'{mm}{dd}{yyyy}'
    
    # set file paths
    in_file_path = Path.joinpath(uiuc_temp_dir, f)
    out_name = Path(f).name
    out_file_path = Path.joinpath(output_dir, iop, out_name)
    print(out_file_path)
    out_file_path.parent.mkdir(parents=True, exist_ok=True)

    shutil.copy2(in_file_path, out_file_path)
```
