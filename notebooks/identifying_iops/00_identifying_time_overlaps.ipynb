```python
all_uiuc_files = list_lidar_files(uiuc_dir,pattern='*.csv', recursive=True)
print(len(all_uiuc_files))
```

```python
all_usc_files = list_lidar_files(usc_dir, pattern='*.txt', recursive=True)
print(len(all_usc_files))
```

```python
usc_summary = pd.DataFrame(columns=['file_name', 'file_path', 'scan_start_time', 'scan_end_time', 'n_bins', 'bin_width_m', 'raw_azimuth_deg', 'raw_zenith_deg'])
for all_usc_file in all_usc_files[:200]:
    usc_meta, usc_profile = read_lidar_file(all_usc_file, source="USC", end_meta=6)
    usc_meta.pop('header_lines')
    usc_meta.pop('dataset_descriptors')
    usc_summary = pd.concat([usc_summary, pd.DataFrame([usc_meta])], ignore_index=True)

    
```
