```python
import tkinter as tk
from matplotlib.figure import Figure
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg, NavigationToolbar2Tk

# 1. Setup the main Tkinter window
root = tk.Tk()
root.title("Embedded Matplotlib in Tkinter")
root.geometry("600x500")

# 2. Create a Matplotlib Figure and Subplot
# Avoid using pyplot (plt.) directly to prevent accidental pop-up windows
fig = Figure(figsize=(5, 4), dpi=100)
ax = fig.add_subplot(111)

# 3. Generate some sample data and plot it
x_data = [1, 2, 3, 4, 5]
y_data = [10, 24, 36, 45, 51]
ax.plot(x_data, y_data, marker='o', color='blue')
ax.set_title("Sample Line Graph")
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")

# 4. Bind the Figure to the Tkinter Canvas widget
canvas = FigureCanvasTkAgg(fig, master=root)
canvas_widget = canvas.get_tk_widget()
canvas_widget.pack(fill=tk.BOTH, expand=True)

# 5. Optional: Add the native Matplotlib interactive toolbar (Zoom, Pan, Save)
toolbar = NavigationToolbar2Tk(canvas, root)
toolbar.update()
toolbar.pack(side=tk.BOTTOM, fill=tk.X)

# 6. Render the plot and start the application loop
canvas.draw()
root.mainloop()
```
